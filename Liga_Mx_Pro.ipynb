```python
"""
========================================================================
 PREDICTOR DE RESULTADOS - LIGA MX (v3 - reordenado y corregido)
 Modelo 1: Regresión de Poisson (ataque/defensa/local)
 Modelo 2: XGBoost -> lambda de Poisson -> matriz de marcadores
 + Datos históricos (GitHub) + datos recientes (ESPN API)
 + Pesos por recencia + Corrección Dixon-Coles + Racha (XGBoost)
========================================================================
Pensado para correr en Google Colab de arriba a abajo, de un jalón
(Runtime -> Run all). El orden de las celdas AQUÍ SÍ es el correcto:
1) descargar histórico  2) agregar datos recientes  3) calcular peso
4) calcular racha  5) entrenar Poisson  6) Dixon-Coles  7) XGBoost
"""

# ------------------------------------------------------------------
# PASO 0: instalar dependencias
# ------------------------------------------------------------------
!pip install xgboost statsmodels -q

import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf
from scipy.stats import poisson
from scipy.optimize import minimize_scalar
import xgboost as xgb
import matplotlib.pyplot as plt
import requests
import time
from datetime import date, timedelta

# ------------------------------------------------------------------
# PASO 1: descargar el dataset histórico de Liga MX (GitHub)
# ------------------------------------------------------------------
!git clone --depth 1 https://github.com/footballcsv/mexico.git /content/ligamx_data

import glob, os

csv_files = glob.glob("/content/ligamx_data/**/*.csv", recursive=True)
csv_files = [f for f in csv_files if os.path.basename(f).lower().startswith("mx")]
print(f"Encontrados {len(csv_files)} archivos CSV de México")

dfs = []
for f in csv_files:
    try:
        tmp = pd.read_csv(f)
        tmp["__source_file"] = f
        dfs.append(tmp)
    except Exception as e:
        print("Salté", f, "->", e)

raw = pd.concat(dfs, ignore_index=True)

# ------------------------------------------------------------------
# PASO 2: limpiar y normalizar el histórico
# ------------------------------------------------------------------
raw = raw[raw["__source_file"].str.contains("mx.1.csv")]  # solo primera división

df = raw.rename(columns={"Team 1": "team1", "Team 2": "team2", "FT": "score"})

mask_valido = df["score"].astype(str).str.match(r"^\d+-\d+$")
df = df[mask_valido]

df[["home_goals", "away_goals"]] = df["score"].str.split("-", expand=True).astype(float)
df = df.dropna(subset=["home_goals", "away_goals", "team1", "team2"])
df["home_goals"] = df["home_goals"].astype(int)
df["away_goals"] = df["away_goals"].astype(int)

df["fecha_dt"] = pd.to_datetime(df["Date"], format="%a %b %d %Y", errors="coerce")

print(f"Partidos históricos limpios: {len(df)}")
print(f"Equipos: {sorted(set(df.team1) | set(df.team2))}")
print(f"Rango de fechas histórico: {df['fecha_dt'].min()} a {df['fecha_dt'].max()}")

# ------------------------------------------------------------------
# PASO 3: traer partidos recientes desde la API pública de ESPN
# (cubre lo que el repo de GitHub no tiene: Apertura 2025 en adelante)
# ------------------------------------------------------------------
def partidos_por_fecha(fecha_str):
    """fecha_str en formato YYYYMMDD"""
    resp = requests.get(
        "https://site.api.espn.com/apis/site/v2/sports/soccer/mex.1/scoreboard",
        params={"dates": fecha_str}
    )
    if resp.status_code != 200:
        return []

    resultados = []
    for ev in resp.json().get("events", []):
        estado = ev.get("status", {}).get("type", {}).get("description")
        if estado != "Full Time":
            continue
        competitors = ev["competitions"][0]["competitors"]
        home = next(c for c in competitors if c["homeAway"] == "home")
        away = next(c for c in competitors if c["homeAway"] == "away")
        resultados.append({
            "fecha_dt": pd.to_datetime(ev["date"]).tz_localize(None),
            "team1": home["team"]["displayName"],
            "team2": away["team"]["displayName"],
            "home_goals": int(home["score"]),
            "away_goals": int(away["score"]),
        })
    return resultados

import os as _os

CACHE_ESPN = "/content/espn_recientes.csv"  # vive en la sesión de Colab; se borra si se reinicia el entorno

if _os.path.exists(CACHE_ESPN):
    manual_df = pd.read_csv(CACHE_ESPN, parse_dates=["fecha_dt"])
    print(f"Usando caché de ESPN ({len(manual_df)} partidos). Borra {CACHE_ESPN} si quieres forzar un fetch nuevo.")
else:
    inicio = df["fecha_dt"].max().date() + timedelta(days=1)  # el día siguiente al último partido del repo
    fin = date.today()

    todos_los_partidos = []
    fecha_actual = inicio
    while fecha_actual <= fin:
        todos_los_partidos.extend(partidos_por_fecha(fecha_actual.strftime("%Y%m%d")))
        fecha_actual += timedelta(days=1)
        time.sleep(0.05)

    manual_df = pd.DataFrame(todos_los_partidos)
    manual_df.to_csv(CACHE_ESPN, index=False)
    print(f"Partidos recientes encontrados en ESPN: {len(manual_df)} (guardados en caché)")

# Homologar nombres de equipo entre ESPN y el dataset histórico
alias_espn_a_dataset = {
    "Atlas": "Atlas Guadalajara",
    "León": "Club León",
    "Querétaro": "Gallos Blancos",
    "Monterrey": "CF Monterrey",
    "Pachuca": "CF Pachuca",
    "América": "CF América",
    "Atlético de San Luis": "Atlético San Luis",
    "FC Juarez": "FC Juárez",
    "Guadalajara": "Deportivo Guadalajara",
    "Tijuana": "Club Tijuana",
    "Necaxa": "Club Necaxa",
    "Puebla": "Puebla FC",
    "Santos": "Santos Laguna",
    "Tigres UANL": "UANL Tigres",
    "Toluca": "Deportivo Toluca",
    # "Atlante" se queda sin mapear a propósito: no existe en el histórico
    # (volvió a primera división después de que termina la cobertura del repo)
}
manual_df["team1"] = manual_df["team1"].replace(alias_espn_a_dataset)
manual_df["team2"] = manual_df["team2"].replace(alias_espn_a_dataset)

desconocidos = (set(manual_df.team1) | set(manual_df.team2)) - (set(df.team1) | set(df.team2))
print("Equipos sin historial previo (se tratan como nuevos):", desconocidos)

manual_df["score"] = manual_df["home_goals"].astype(str) + "-" + manual_df["away_goals"].astype(str)

# ------------------------------------------------------------------
# PASO 4: unir histórico + reciente ANTES de calcular peso y racha
# ------------------------------------------------------------------
antes = len(df)
df = pd.concat([df, manual_df], ignore_index=True, sort=False)
print(f"Partidos antes de unir: {antes}  |  después: {len(df)}")
print(f"Rango de fechas final: {df['fecha_dt'].min()} a {df['fecha_dt'].max()}")

# ------------------------------------------------------------------
# PASO 5: peso por recencia (ya con el dataset completo)
# ------------------------------------------------------------------
fecha_referencia = df["fecha_dt"].max()
df["dias_atras"] = (fecha_referencia - df["fecha_dt"]).dt.days

xi = 0.001
df["peso"] = np.exp(-xi * df["dias_atras"])
print(df[["fecha_dt", "dias_atras", "peso"]].describe())

# ------------------------------------------------------------------
# PASO 6: racha de los últimos 5 partidos (sin fuga de información)
# ------------------------------------------------------------------
df_sorted = df.sort_values("fecha_dt").reset_index(drop=True)

def puntos(gf, gc):
    if gf > gc:
        return 3
    elif gf == gc:
        return 1
    return 0

historial = []
for _, r in df_sorted.iterrows():
    historial.append({"team": r.team1, "fecha": r.fecha_dt,
                       "puntos": puntos(r.home_goals, r.away_goals),
                       "gf": r.home_goals, "gc": r.away_goals})
    historial.append({"team": r.team2, "fecha": r.fecha_dt,
                       "puntos": puntos(r.away_goals, r.home_goals),
                       "gf": r.away_goals, "gc": r.home_goals})

historial_df = pd.DataFrame(historial).sort_values(["team", "fecha"]).reset_index(drop=True)

for col in ["puntos", "gf", "gc"]:
    historial_df[f"racha_{col}"] = (
        historial_df.groupby("team")[col]
        .transform(lambda s: s.shift(1).rolling(5, min_periods=1).mean())
    )
    historial_df[f"racha_{col}"] = historial_df[f"racha_{col}"].fillna(historial_df[col].mean())

print(f"Total registros de historial: {len(historial_df)}")

# ------------------------------------------------------------------
# MODELO 1: Regresión de Poisson (statsmodels GLM), con pesos
# ------------------------------------------------------------------
home_rows = df[["team1", "team2", "home_goals", "peso"]].rename(
    columns={"team1": "team", "team2": "opponent", "home_goals": "goals"})
home_rows["is_home"] = 1

away_rows = df[["team2", "team1", "away_goals", "peso"]].rename(
    columns={"team2": "team", "team1": "opponent", "away_goals": "goals"})
away_rows["is_home"] = 0

long_df = pd.concat([home_rows, away_rows], ignore_index=True)

poisson_model = smf.glm(
    formula="goals ~ is_home + team + opponent",
    data=long_df,
    family=sm.families.Poisson(),
    freq_weights=long_df["peso"]
).fit()

print(poisson_model.summary())

# ------------------------------------------------------------------
# CORRECCIÓN DIXON-COLES (marcadores bajos: 0-0, 1-0, 0-1, 1-1)
# ------------------------------------------------------------------
def tau_dixon_coles(x, y, lam, mu, rho):
    if x == 0 and y == 0:
        return 1 - (lam * mu * rho)
    elif x == 0 and y == 1:
        return 1 + (lam * rho)
    elif x == 1 and y == 0:
        return 1 + (mu * rho)
    elif x == 1 and y == 1:
        return 1 - rho
    return 1.0

def neg_log_lik_rho(rho, data, model):
    ll = 0.0
    for _, r in data.iterrows():
        lam = model.predict(pd.DataFrame({
            "team": [r.team1], "opponent": [r.team2], "is_home": [1]
        })).iloc[0]
        mu = model.predict(pd.DataFrame({
            "team": [r.team2], "opponent": [r.team1], "is_home": [0]
        })).iloc[0]
        x, y = int(r.home_goals), int(r.away_goals)
        tau = max(tau_dixon_coles(x, y, lam, mu, rho), 1e-10)
        ll += poisson.logpmf(x, lam) + poisson.logpmf(y, mu) + np.log(tau)
    return -ll

muestra = df.sample(min(500, len(df)), random_state=42)
resultado_rho = minimize_scalar(
    neg_log_lik_rho, bounds=(-0.3, 0.3), method="bounded",
    args=(muestra, poisson_model)
)
rho = resultado_rho.x
print(f"Rho estimado (Dixon-Coles): {rho:.4f}")

def matriz_con_dixon_coles(lam, mu, rho, max_goals=6):
    home_probs = poisson.pmf(np.arange(max_goals + 1), lam)
    away_probs = poisson.pmf(np.arange(max_goals + 1), mu)
    matrix = np.outer(home_probs, away_probs)
    for x in range(2):
        for y in range(2):
            matrix[x, y] *= tau_dixon_coles(x, y, lam, mu, rho)
    matrix /= matrix.sum()
    return matrix

def predict_poisson(home_team, away_team, max_goals=6):
    lam_home = poisson_model.predict(pd.DataFrame({
        "team": [home_team], "opponent": [away_team], "is_home": [1]
    })).iloc[0]
    lam_away = poisson_model.predict(pd.DataFrame({
        "team": [away_team], "opponent": [home_team], "is_home": [0]
    })).iloc[0]
    matrix = matriz_con_dixon_coles(lam_home, lam_away, rho, max_goals)
    return matrix, lam_home, lam_away

# ------------------------------------------------------------------
# long_df2: igual que long_df, pero con fecha + racha pegada (para XGBoost)
# ------------------------------------------------------------------
home_rows2 = df_sorted[["team1", "team2", "home_goals", "peso", "fecha_dt"]].rename(
    columns={"team1": "team", "team2": "opponent", "home_goals": "goals", "fecha_dt": "fecha"})
home_rows2["is_home"] = 1

away_rows2 = df_sorted[["team2", "team1", "away_goals", "peso", "fecha_dt"]].rename(
    columns={"team2": "team", "team1": "opponent", "away_goals": "goals", "fecha_dt": "fecha"})
away_rows2["is_home"] = 0

long_df2 = pd.concat([home_rows2, away_rows2], ignore_index=True)

historial_sin_dup = historial_df.drop_duplicates(subset=["team", "fecha"], keep="first")
long_df2 = long_df2.merge(
    historial_sin_dup[["team", "fecha", "racha_puntos", "racha_gf", "racha_gc"]],
    on=["team", "fecha"], how="left"
)

print(f"long_df2 filas: {len(long_df2)}  (debería ser {len(df)*2})")
print(f"Nulos tras merge: {long_df2['racha_puntos'].isna().sum()}")

# ------------------------------------------------------------------
# MODELO 2: XGBoost -> lambda esperada -> Poisson + Dixon-Coles
# ------------------------------------------------------------------
features_df = pd.get_dummies(
    long_df2[["team", "opponent", "is_home", "racha_puntos", "racha_gf", "racha_gc"]],
    columns=["team", "opponent"]
)
target = long_df2["goals"]

xgb_model = xgb.XGBRegressor(
    objective="count:poisson",
    n_estimators=300,
    max_depth=3,
    learning_rate=0.05,
)
xgb_model.fit(features_df, target, sample_weight=long_df2["peso"])

def predict_xgboost(home_team, away_team, max_goals=6):
    def make_row(team, opponent, is_home):
        row = pd.DataFrame(np.zeros((1, features_df.shape[1])),
                            columns=features_df.columns)
        row["is_home"] = is_home
        col_team = f"team_{team}"
        col_opp = f"opponent_{opponent}"
        if col_team in row.columns:
            row[col_team] = 1
        if col_opp in row.columns:
            row[col_opp] = 1

        historial_equipo = historial_df[historial_df.team == team]
        if len(historial_equipo) > 0:
            ultima_racha = historial_equipo.iloc[-1]
            row["racha_puntos"] = ultima_racha["puntos"]
            row["racha_gf"] = ultima_racha["gf"]
            row["racha_gc"] = ultima_racha["gc"]
        return row

    lam_home = xgb_model.predict(make_row(home_team, away_team, 1))[0]
    lam_away = xgb_model.predict(make_row(away_team, home_team, 0))[0]
    matrix = matriz_con_dixon_coles(lam_home, lam_away, rho, max_goals)
    return matrix, lam_home, lam_away

# ------------------------------------------------------------------
# REPORTAR resultados de un partido
# ------------------------------------------------------------------
def summarize(matrix, home_team, away_team, lam_home, lam_away, label):
    p_home = np.tril(matrix, -1).sum()
    p_draw = np.trace(matrix)
    p_away = np.triu(matrix, 1).sum()
    print(f"\n=== {label}: {home_team} (local) vs {away_team} (visita) ===")
    print(f"Goles esperados: {home_team}={lam_home:.2f}  {away_team}={lam_away:.2f}")
    print(f"P(gana {home_team}) = {p_home:.1%}")
    print(f"P(empate)          = {p_draw:.1%}")
    print(f"P(gana {away_team}) = {p_away:.1%}")

    top_idx = np.dstack(np.unravel_index(np.argsort(-matrix.ravel())[:5], matrix.shape))[0]
    print("Marcadores más probables:")
    for hg, ag in top_idx:
        print(f"  {home_team} {hg}-{ag} {away_team}: {matrix[hg, ag]:.1%}")

    plt.figure(figsize=(6, 5))
    plt.imshow(matrix[:6, :6], cmap="viridis")
    plt.colorbar(label="Probabilidad")
    plt.xlabel(f"Goles {away_team}")
    plt.ylabel(f"Goles {home_team}")
    plt.title(f"{label}: {home_team} vs {away_team}")
    plt.xticks(range(6))
    plt.yticks(range(6))
    plt.show()

# ------------------------------------------------------------------
# PREDICE VARIOS PARTIDOS DE UN JALÓN
# ------------------------------------------------------------------
partidos = [
    ("Club Necaxa", "Cruz Azul"),
    ("Club Tijuana", "Pumas UNAM"),
    ("Atlas Guadalajara", "Gallos Blancos"),
    ("CF Pachuca", "Deportivo Guadalajara"),
    ("CF América", "Puebla FC"),
    ("Santos Laguna", "UANL Tigres"),
    ("Deportivo Toluca", "FC Juárez"),
    ("CF Monterrey", "Atlético San Luis"),
    ("Atlético San Luis", "Mazatlán FC"),
    ("Santos Laguna", "Cruz Azul"),
]

for home_team, away_team in partidos:
    matrix_p, lh_p, la_p = predict_poisson(home_team, away_team)
    summarize(matrix_p, home_team, away_team, lh_p, la_p, "Poisson (statsmodels) + Dixon-Coles")

    matrix_x, lh_x, la_x = predict_xgboost(home_team, away_team)
    summarize(matrix_x, home_team, away_team, lh_x, la_x, "XGBoost + Poisson + Dixon-Coles")
```
