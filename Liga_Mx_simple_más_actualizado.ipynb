```python
"""
========================================================================
 PREDICTOR DE RESULTADOS - LIGA MX (v_original_actualizado)
 Mismo esqueleto que el script "original" (Poisson simple + XGBoost
 simple, SIN pesos por recencia, SIN Dixon-Coles, SIN racha) pero
 con el histórico completo hasta HOY (GitHub + ESPN).

 Sirve como punto de comparación contra el v3: aquí un partido de
 2018 pesa exactamente igual que uno de la semana pasada. Si este
 modelo y el v3 coinciden en un partido, hay más confianza. Si
 difieren mucho, es señal de que un equipo está en una racha
 inusual respecto a su promedio histórico completo.
========================================================================
Pensado para correr en Google Colab de arriba a abajo (Runtime -> Run all).
"""

# ------------------------------------------------------------------
# PASO 0: instalar dependencias
# ------------------------------------------------------------------
!pip install xgboost statsmodels -q

import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf
from scipy.stats import poisson
import xgboost as xgb
import matplotlib.pyplot as plt
import requests
import time
from datetime import date, timedelta

# ------------------------------------------------------------------
# PASO 1: descargar el histórico de Liga MX (GitHub, hasta ~2024-25)
# ------------------------------------------------------------------
!git clone --depth 1 https://github.com/footballcsv/mexico.git /content/ligamx_data

import glob, os

csv_files = glob.glob("/content/ligamx_data/**/*.csv", recursive=True)
csv_files = [f for f in csv_files if os.path.basename(f).lower().startswith("mx")]
print(f"Encontrados {len(csv_files)} archivos CSV de México")

dfs = []
for f in csv_files:
    try:
        tmp = pd.read_csv(f)
        tmp["__source_file"] = f
        dfs.append(tmp)
    except Exception as e:
        print("Salté", f, "->", e)

raw = pd.concat(dfs, ignore_index=True)

# ------------------------------------------------------------------
# PASO 2: limpiar y normalizar el histórico
# ------------------------------------------------------------------
raw = raw[raw["__source_file"].str.contains("mx.1.csv")]  # solo primera división

df = raw.rename(columns={"Team 1": "team1", "Team 2": "team2", "FT": "score"})

mask_valido = df["score"].astype(str).str.match(r"^\d+-\d+$")
df = df[mask_valido]

df[["home_goals", "away_goals"]] = df["score"].str.split("-", expand=True).astype(float)
df = df.dropna(subset=["home_goals", "away_goals", "team1", "team2"])
df["home_goals"] = df["home_goals"].astype(int)
df["away_goals"] = df["away_goals"].astype(int)

df["fecha_dt"] = pd.to_datetime(df["Date"], format="%a %b %d %Y", errors="coerce")

print(f"Partidos históricos limpios: {len(df)}")
print(f"Rango de fechas histórico: {df['fecha_dt'].min()} a {df['fecha_dt'].max()}")

# ------------------------------------------------------------------
# PASO 3: traer partidos recientes desde la API pública de ESPN
# (mismo mecanismo que el v3, con su propio caché independiente)
# ------------------------------------------------------------------
def partidos_por_fecha(fecha_str):
    resp = requests.get(
        "https://site.api.espn.com/apis/site/v2/sports/soccer/mex.1/scoreboard",
        params={"dates": fecha_str}
    )
    if resp.status_code != 200:
        return []
    resultados = []
    for ev in resp.json().get("events", []):
        estado = ev.get("status", {}).get("type", {}).get("description")
        if estado != "Full Time":
            continue
        competitors = ev["competitions"][0]["competitors"]
        home = next(c for c in competitors if c["homeAway"] == "home")
        away = next(c for c in competitors if c["homeAway"] == "away")
        resultados.append({
            "fecha_dt": pd.to_datetime(ev["date"]).tz_localize(None),
            "team1": home["team"]["displayName"],
            "team2": away["team"]["displayName"],
            "home_goals": int(home["score"]),
            "away_goals": int(away["score"]),
        })
    return resultados

CACHE_ESPN = "/content/espn_recientes_v_original.csv"  # caché propio, distinto al del v3

if os.path.exists(CACHE_ESPN):
    manual_df = pd.read_csv(CACHE_ESPN, parse_dates=["fecha_dt"])
    print(f"Usando caché de ESPN ({len(manual_df)} partidos). Borra {CACHE_ESPN} si quieres forzar un fetch nuevo.")
else:
    inicio = df["fecha_dt"].max().date() + timedelta(days=1)
    fin = date.today()

    todos_los_partidos = []
    fecha_actual = inicio
    while fecha_actual <= fin:
        todos_los_partidos.extend(partidos_por_fecha(fecha_actual.strftime("%Y%m%d")))
        fecha_actual += timedelta(days=1)
        time.sleep(0.05)

    manual_df = pd.DataFrame(todos_los_partidos)
    manual_df.to_csv(CACHE_ESPN, index=False)
    print(f"Partidos recientes encontrados en ESPN: {len(manual_df)} (guardados en caché)")

alias_espn_a_dataset = {
    "Atlas": "Atlas Guadalajara",
    "León": "Club León",
    "Querétaro": "Gallos Blancos",
    "Monterrey": "CF Monterrey",
    "Pachuca": "CF Pachuca",
    "América": "CF América",
    "Atlético de San Luis": "Atlético San Luis",
    "FC Juarez": "FC Juárez",
    "Guadalajara": "Deportivo Guadalajara",
    "Tijuana": "Club Tijuana",
    "Necaxa": "Club Necaxa",
    "Puebla": "Puebla FC",
    "Santos": "Santos Laguna",
    "Tigres UANL": "UANL Tigres",
    "Toluca": "Deportivo Toluca",
    # "Atlante" se deja sin mapear: no existe en el histórico
}
manual_df["team1"] = manual_df["team1"].replace(alias_espn_a_dataset)
manual_df["team2"] = manual_df["team2"].replace(alias_espn_a_dataset)
manual_df["score"] = manual_df["home_goals"].astype(str) + "-" + manual_df["away_goals"].astype(str)

# ------------------------------------------------------------------
# PASO 4: unir histórico + reciente. SIN pesos: todo pesa igual.
# ------------------------------------------------------------------
antes = len(df)
df = pd.concat([df, manual_df], ignore_index=True, sort=False)
print(f"Partidos antes de unir: {antes}  |  después: {len(df)}")
print(f"Rango de fechas final: {df['fecha_dt'].min()} a {df['fecha_dt'].max()}")
print(f"Equipos: {sorted(set(df.team1) | set(df.team2))}")

# ------------------------------------------------------------------
# MODELO 1: Regresión de Poisson SIMPLE (sin pesos, sin Dixon-Coles)
# ------------------------------------------------------------------
home_rows = df[["team1", "team2", "home_goals"]].rename(
    columns={"team1": "team", "team2": "opponent", "home_goals": "goals"})
home_rows["is_home"] = 1

away_rows = df[["team2", "team1", "away_goals"]].rename(
    columns={"team2": "team", "team1": "opponent", "away_goals": "goals"})
away_rows["is_home"] = 0

long_df = pd.concat([home_rows, away_rows], ignore_index=True)

poisson_model = smf.glm(
    formula="goals ~ is_home + team + opponent",
    data=long_df,
    family=sm.families.Poisson()
).fit()

print(poisson_model.summary())

def predict_poisson(home_team, away_team, max_goals=6):
    lam_home = poisson_model.predict(pd.DataFrame({
        "team": [home_team], "opponent": [away_team], "is_home": [1]
    })).iloc[0]
    lam_away = poisson_model.predict(pd.DataFrame({
        "team": [away_team], "opponent": [home_team], "is_home": [0]
    })).iloc[0]
    home_probs = poisson.pmf(np.arange(max_goals + 1), lam_home)
    away_probs = poisson.pmf(np.arange(max_goals + 1), lam_away)
    matrix = np.outer(home_probs, away_probs)
    return matrix, lam_home, lam_away

# ------------------------------------------------------------------
# MODELO 2: XGBoost SIMPLE (sin racha, sin pesos)
# ------------------------------------------------------------------
features_df = pd.get_dummies(long_df[["team", "opponent", "is_home"]],
                              columns=["team", "opponent"])
target = long_df["goals"]

xgb_model = xgb.XGBRegressor(
    objective="count:poisson",
    n_estimators=300,
    max_depth=3,
    learning_rate=0.05,
)
xgb_model.fit(features_df, target)

def predict_xgboost(home_team, away_team, max_goals=6):
    def make_row(team, opponent, is_home):
        row = pd.DataFrame(np.zeros((1, features_df.shape[1])),
                            columns=features_df.columns)
        row["is_home"] = is_home
        col_team = f"team_{team}"
        col_opp = f"opponent_{opponent}"
        if col_team in row.columns:
            row[col_team] = 1
        if col_opp in row.columns:
            row[col_opp] = 1
        return row

    lam_home = xgb_model.predict(make_row(home_team, away_team, 1))[0]
    lam_away = xgb_model.predict(make_row(away_team, home_team, 0))[0]
    home_probs = poisson.pmf(np.arange(max_goals + 1), lam_home)
    away_probs = poisson.pmf(np.arange(max_goals + 1), lam_away)
    matrix = np.outer(home_probs, away_probs)
    return matrix, lam_home, lam_away

# ------------------------------------------------------------------
# REPORTAR resultados de un partido
# ------------------------------------------------------------------
def summarize(matrix, home_team, away_team, lam_home, lam_away, label):
    p_home = np.tril(matrix, -1).sum()
    p_draw = np.trace(matrix)
    p_away = np.triu(matrix, 1).sum()
    print(f"\n=== {label}: {home_team} (local) vs {away_team} (visita) ===")
    print(f"Goles esperados: {home_team}={lam_home:.2f}  {away_team}={lam_away:.2f}")
    print(f"P(gana {home_team}) = {p_home:.1%}")
    print(f"P(empate)          = {p_draw:.1%}")
    print(f"P(gana {away_team}) = {p_away:.1%}")

    top_idx = np.dstack(np.unravel_index(np.argsort(-matrix.ravel())[:5], matrix.shape))[0]
    print("Marcadores más probables:")
    for hg, ag in top_idx:
        print(f"  {home_team} {hg}-{ag} {away_team}: {matrix[hg, ag]:.1%}")

    plt.figure(figsize=(6, 5))
    plt.imshow(matrix[:6, :6], cmap="viridis")
    plt.colorbar(label="Probabilidad")
    plt.xlabel(f"Goles {away_team}")
    plt.ylabel(f"Goles {home_team}")
    plt.title(f"{label}: {home_team} vs {away_team}")
    plt.xticks(range(6))
    plt.yticks(range(6))
    plt.show()

# ------------------------------------------------------------------
# PREDICE VARIOS PARTIDOS DE UN JALÓN
# ------------------------------------------------------------------
partidos = [
    ("Club Necaxa", "Cruz Azul"),
    ("Club Tijuana", "Pumas UNAM"),
    ("Atlas Guadalajara", "Gallos Blancos"),
    ("CF Pachuca", "Deportivo Guadalajara"),
    ("CF América", "Puebla FC"),
    ("Santos Laguna", "UANL Tigres"),
    ("Deportivo Toluca", "FC Juárez"),
    ("CF Monterrey", "Atlético San Luis"),
    ("Atlético San Luis", "Mazatlán FC"),
    ("Santos Laguna", "Cruz Azul"),
]

for home_team, away_team in partidos:
    matrix_p, lh_p, la_p = predict_poisson(home_team, away_team)
    summarize(matrix_p, home_team, away_team, lh_p, la_p, "Poisson simple (todo el histórico, sin pesos)")

    matrix_x, lh_x, la_x = predict_xgboost(home_team, away_team)
    summarize(matrix_x, home_team, away_team, lh_x, la_x, "XGBoost simple (todo el histórico, sin pesos)")
```
